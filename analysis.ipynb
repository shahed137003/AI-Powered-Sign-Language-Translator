```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
output_dir = "visuals"#output folder
os.makedirs(output_dir, exist_ok=True)
```

```python
csv_path = "gloss_counts.csv"
df = pd.read_csv(csv_path)
df = df.sort_values(by="count", ascending=False).reset_index(drop=True)
df.head()
```

```python
# Basic Stats
print("Total glosses:", len(df))
print("Total instances:", df["count"].sum())
```

```python
rare = df[df["count"] < 5]
medium = df[(df["count"] >= 5) & (df["count"] <= 20)]
frequent = df[df["count"] > 20]
```

```python
# 1. Top 20 Glosses
top_df = df.head(20)
plt.figure(figsize=(12,6))
plt.bar(top_df["gloss"], top_df["count"])
plt.xticks(rotation=45, ha="right")
plt.title("Top 20 Most Frequent Glosses")
plt.xlabel("Gloss")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "top_20_glosses.png"), dpi=300)
plt.show()
plt.close()
```

```python
# Least 10 Glosses
bottom10 = df.tail(10)

plt.figure(figsize=(12,6))
plt.bar(bottom10["gloss"], bottom10["count"])
plt.xticks(rotation=45, ha="right")
plt.title("Least 10 Frequent Glosses")
plt.xlabel("Gloss")
plt.ylabel("Count")
plt.tight_layout()

plt.savefig(os.path.join(output_dir, "least_10_glosses.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 2. Distribution Histogram
plt.figure(figsize=(12,6))

plt.hist(df["count"], bins=25, edgecolor="black")
plt.title("Distribution of Gloss Instance Counts")
plt.xlabel("Instances per Gloss")
plt.ylabel("Number of Glosses")
plt.tight_layout()

plt.savefig(os.path.join(output_dir, "distribution_histogram.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 3. Log Scale Histogram
plt.figure(figsize=(12,6))
plt.hist(df["count"], bins=25, edgecolor="black")
plt.yscale("log")
plt.title("Gloss Distribution (Log Scale)")
plt.xlabel("Instances per Gloss")
plt.ylabel("Log Frequency")
plt.tight_layout()

plt.savefig(os.path.join(output_dir, "distribution_log_scale.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 4. Box Plot
plt.figure(figsize=(12,6))

plt.boxplot(df["count"], vert=False)
plt.title("Gloss Count Distribution (Outliers View)")
plt.xlabel("Instance Count")
plt.tight_layout()

plt.savefig(os.path.join(output_dir, "boxplot.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 6. Class Breakdown
labels = ["Rare (<5)", "Medium (5–20)", "Frequent (>20)"]
sizes = [len(rare), len(medium), len(frequent)]

plt.figure(figsize=(12,6))

plt.bar(labels, sizes)
plt.title("Class Frequency Breakdown")
plt.ylabel("Number of Glosses")
plt.tight_layout()

plt.savefig(os.path.join(output_dir, "class_breakdown.png"), dpi=300)
plt.show()
plt.close()
```
