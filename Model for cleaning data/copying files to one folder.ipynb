```python
import shutil
import re
from pathlib import Path

SRC_DIR = Path("clean_dataset")
DST_DIR = Path("all_clean_dataset")
DST_DIR.mkdir(parents=True, exist_ok=True)

for folder in SRC_DIR.iterdir():
    if not folder.is_dir():
        continue

    # extract ONLY _v1, _v2...
    match = re.search(r"(_v\d+)", folder.name)
    variant = match.group(1) if match else ""

    for file in folder.glob("*.npy"):
        parts = file.stem.split()

        if len(parts) < 2:
            continue

        # detect numbers
        leading_num = parts[0] if parts[0].isdigit() else ""
        trailing_num = parts[-1] if parts[-1].isdigit() else ""

        # remove numbers from word
        word_parts = [p for p in parts if not p.isdigit()]

        # CLEANING STEP (IMPORTANT)
        cleaned_words = []
        for w in word_parts:
            w = w.replace("_", " ")
            w = w.replace("-", " ")
            cleaned_words.extend(w.split())

        word = " ".join(cleaned_words).strip()

        # inject variant ONLY here
        if variant:
            word = word + variant

        # rebuild filename
        if leading_num and trailing_num:
            new_name = f"{leading_num} {word} {trailing_num}.npy"
        elif trailing_num:
            new_name = f"{word} {trailing_num}.npy"
        else:
            new_name = f"{word}.npy"

        shutil.copy(file, DST_DIR / new_name)

print("DONE CLEANING")
```
