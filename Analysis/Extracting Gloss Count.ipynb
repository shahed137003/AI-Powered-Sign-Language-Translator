```python
import os
import pandas as pd
from collections import defaultdict

DATASET_PATH =r"D:\100 words sample\100words_cleaned_not_preprocessed_no_mask\cleaned_landmarks_100words"
# Dictionary to count instances
gloss_counts = defaultdict(int)

for file in os.listdir(DATASET_PATH):
    if file.endswith(".npy"):
        name = file.replace(".npy", "")

        parts = name.split()

        # Check if last part is a number (instance index)
        if parts[-1].isdigit():
            gloss = " ".join(parts[:-1])
        else:
            gloss = name  # whole name is gloss

        # Convert to Title Case
        gloss = gloss.title()

        # Count instance
        gloss_counts[gloss] += 1

# Convert to DataFrame
df = pd.DataFrame(list(gloss_counts.items()), columns=["Gloss", "Instances"])

# Sort descending
df = df.sort_values(by="Instances", ascending=False)

# Save to CSV
output_path = "100_words_data_gloss_counts.csv"
df.to_csv(output_path, index=False)

print(f"Saved to {output_path}")
print(df.head())
```
