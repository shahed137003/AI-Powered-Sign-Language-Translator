```python
class_map = np.load(
    r"E:\500 new with cleaning\500_final_dataset_modified\class_map.npy",
    allow_pickle=True
)

print(class_map)
```

```python
low_glosses = set(df[df['f1-score'] < 0.7]['Gloss'])

matching_indices = [
    class_map[gloss]
    for gloss in low_glosses
    if gloss in class_map
]

print(matching_indices)
```

```python
from pathlib import Path
import numpy as np

root = Path(r"E:\500 new with cleaning\500_final_dataset_modified")
remove_indices = set(matching_indices)

deleted = 0

for split in ["train", "val", "test"]:
    split_dir = root / split

    for file in split_dir.glob("*.npz"):
        try:
            with np.load(file) as data:   # 👈 IMPORTANT FIX
                y = int(data["y"])

            if y in remove_indices:
                file.unlink()
                deleted += 1

        except Exception as e:
            print(f"Error: {file} -> {e}")

print(f"Deleted {deleted} files")
```

```python
from pathlib import Path
import numpy as np

root = Path(r"E:\500 new with cleaning\500_final_dataset_modified")

all_classes = set()

for split in ["train", "val", "test"]:
    split_dir = root / split

    for file in split_dir.glob("*.npz"):
        with np.load(file) as data:
            all_classes.add(int(data["y"]))

print("Number of remaining classes:", len(all_classes))
```
