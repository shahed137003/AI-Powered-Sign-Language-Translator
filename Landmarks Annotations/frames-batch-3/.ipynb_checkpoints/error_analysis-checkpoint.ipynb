```python
import os
import json
import cv2
import numpy as np
import matplotlib.pyplot as plt
import xml.etree.ElementTree as ET
import warnings
warnings.filterwarnings('ignore')

THRESHOLDS = [10, 20]

MP_JOINTS = [
    "WRIST", "THUMB_CMC", "THUMB_MCP", "THUMB_IP", "THUMB_TIP",
    "INDEX_FINGER_MCP", "INDEX_FINGER_PIP", "INDEX_FINGER_DIP", "INDEX_FINGER_TIP",
    "MIDDLE_FINGER_MCP", "MIDDLE_FINGER_PIP", "MIDDLE_FINGER_DIP", "MIDDLE_FINGER_TIP",
    "RING_FINGER_MCP", "RING_FINGER_PIP", "RING_FINGER_DIP", "RING_FINGER_TIP",
    "PINKY_MCP", "PINKY_PIP", "PINKY_DIP", "PINKY_TIP"
]
JOINT_TO_IDX = {name: i for i, name in enumerate(MP_JOINTS)}

FINGER_GROUPS = {
    "thumb": MP_JOINTS[1:5],
    "index": MP_JOINTS[5:9],
    "middle": MP_JOINTS[9:13],
    "ring": MP_JOINTS[13:17],
    "pinky": MP_JOINTS[17:21]
}

HAND_CONNECTIONS = [
    (0,1),(1,2),(2,3),(3,4),
    (0,5),(5,6),(6,7),(7,8),
    (0,9),(9,10),(10,11),(11,12),
    (0,13),(13,14),(14,15),(15,16),
    (0,17),(17,18),(18,19),(19,20)
]

def draw_skeleton(ax, joints, connections, color='green', marker='o', label=None):
    xs, ys = [], []
    for pt in joints.values():
        xs.append(pt[0])
        ys.append(pt[1])
    ax.scatter(xs, ys, c=color, marker=marker, label=label, s=40, edgecolors='black')
    for s, e in connections:
        s_name, e_name = MP_JOINTS[s], MP_JOINTS[e]
        if s_name in joints and e_name in joints:
            ax.plot([joints[s_name][0], joints[e_name][0]],
                    [joints[s_name][1], joints[e_name][1]],
                    c=color, linewidth=2)

def init_global_stats():
    return {
        "total_joints": 0,
        "all_errors": [],
        "correct": {t: 0 for t in THRESHOLDS},
        "per_joint": {j: {"total":0, "correct":0} for j in MP_JOINTS},
        "total_hands": 0,
        "detected_hands": 0,
        "missed_hands": 0,
        "swap_errors": 0,
        "per_hand_detection": {"left": {"total":0, "detected":0}, "right":{"total":0,"detected":0}}
    }

def run_analysis(batch_path, stats, global_mistakes):
    ERROR_JSON = os.path.join(batch_path, 'pixel_errors_by_joint.json')
    MP_JSON = os.path.join(batch_path, 'mediapipe_landmarks.json')
    XML_FILE = os.path.join(batch_path, 'annotations.xml')
    IMAGE_DIR = os.path.join(batch_path, 'frames')

    GLOBAL_IMG_ROOT = "mistake_images"
    batch_name = os.path.basename(batch_path)
    SAVE_DIR = os.path.join(GLOBAL_IMG_ROOT, batch_name)
    os.makedirs(SAVE_DIR, exist_ok=True)

    with open(ERROR_JSON) as f:
        error_data = json.load(f)
    with open(MP_JSON) as f:
        mp_data = json.load(f)

    root = ET.parse(XML_FILE).getroot()

    for image_tag in root.findall('image'):
        frame_name = os.path.basename(image_tag.get('name'))
        width = int(image_tag.get('width'))
        height = int(image_tag.get('height'))

        img_path = os.path.join(IMAGE_DIR, frame_name)
        img = cv2.imread(img_path)
        if img is None:
            img = np.zeros((height, width, 3), dtype=np.uint8)

        fig, ax = plt.subplots(figsize=(width/100, height/100), dpi=100)
        ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        ax.axis('off')

        frame_has_error = False

        for skeleton in image_tag.findall('skeleton'):
            hand = skeleton.get('label')
            stats["total_hands"] += 1
            stats["per_hand_detection"][hand.replace("_hand","")]["total"] += 1

            gt_bad, mp_bad = {}, {}

            if hand in mp_data.get(frame_name, {}) and hand in error_data.get(frame_name, {}):
                stats["detected_hands"] += 1
                stats["per_hand_detection"][hand.replace("_hand","")]["detected"] +=1

                for p in skeleton.findall('points'):
                    joint = p.get('label')
                    gx, gy = map(float, p.get('points').split(','))
                    stats["total_joints"] += 1

                    if joint not in error_data[frame_name][hand]:
                        error_val = 999  # large error if joint missing
                    else:
                        error_val = error_data[frame_name][hand][joint]

                    stats["all_errors"].append(error_val)
                    stats["per_joint"][joint]["total"] += 1

                    for t in THRESHOLDS:
                        if error_val <= t:
                            stats["correct"][t] += 1

                    if error_val <= 10:
                        stats["per_joint"][joint]["correct"] += 1
                    else:
                        frame_has_error = True
                        mp_pt = mp_data[frame_name][hand][JOINT_TO_IDX[joint]]
                        mx, my = mp_pt['x']*width, mp_pt['y']*height
                        gt_bad[joint] = (gx, gy)
                        mp_bad[joint] = (mx, my)

            else:
                # Hand completely missed: count all 21 joints as wrong
                stats["missed_hands"] += 1
                stats["swap_errors"] += 1  # optional swap error counting
                for joint in MP_JOINTS:
                    stats["total_joints"] += 1
                    stats["all_errors"].append(999)
                    stats["per_joint"][joint]["total"] +=1

                frame_has_error = True

            if gt_bad and mp_bad:
                draw_skeleton(ax, gt_bad, HAND_CONNECTIONS, 'green', 'o', f'GT {hand}')
                draw_skeleton(ax, mp_bad, HAND_CONNECTIONS, 'red', 'x', f'MP {hand}')
                for j in gt_bad:
                    if j in mp_bad:
                        ax.plot([gt_bad[j][0], mp_bad[j][0]],
                                [gt_bad[j][1], mp_bad[j][1]],
                                c='blue', linewidth=1)

        if frame_has_error:
            plt.tight_layout()
            plt.savefig(os.path.join(SAVE_DIR, frame_name))
        plt.close(fig)

def compute_final_metrics(stats):
    total_joints = stats["total_joints"]
    errors = np.array(stats["all_errors"])
    total_hands = stats["total_hands"]
    detected_hands = stats["detected_hands"]
    missed_hands = stats["missed_hands"]

    metrics = {
        "hand_detection_rate": detected_hands / total_hands if total_hands else 0,
        "miss_rate": missed_hands / total_hands if total_hands else 0,
        "swap_error_rate": stats.get("swap_errors",0)/total_hands if total_hands else 0,
        "MPJPE": float(np.mean(errors[errors<999])) if len(errors) else 0,
        "total_joints_evaluated": total_joints,
        "per_hand_detection": {
            "left": {
                "total": stats["per_hand_detection"]["left"]["total"],
                "detected": stats["per_hand_detection"]["left"]["detected"],
                "accuracy": (stats["per_hand_detection"]["left"]["detected"]/stats["per_hand_detection"]["left"]["total"]
                             if stats["per_hand_detection"]["left"]["total"] else 0)
            },
            "right": {
                "total": stats["per_hand_detection"]["right"]["total"],
                "detected": stats["per_hand_detection"]["right"]["detected"],
                "accuracy": (stats["per_hand_detection"]["right"]["detected"]/stats["per_hand_detection"]["right"]["total"]
                             if stats["per_hand_detection"]["right"]["total"] else 0)
            }
        }
    }

    for t in THRESHOLDS:
        metrics[f"PCK@{t}"] = {
            "correct": stats["correct"][t],
            "total": total_joints,
            "accuracy": stats["correct"][t]/total_joints if total_joints else 0
        }

    per_joint_acc = {}
    for j, d in stats["per_joint"].items():
        per_joint_acc[j] = {
            "correct": d["correct"],
            "total": d["total"],
            "accuracy": d["correct"]/d["total"] if d["total"]>0 else 0
        }
    metrics["per_joint_accuracy"] = per_joint_acc

    finger_acc = {}
    for f, joints in FINGER_GROUPS.items():
        vals = [per_joint_acc[j]["accuracy"] for j in joints if per_joint_acc[j]["total"]>0]
        if vals:
            finger_acc[f] = float(np.mean(vals))
    metrics["per_finger_accuracy"] = finger_acc

    # error distribution
    if len(errors):
        metrics["error_distribution"] = {
            "5-10": float(np.mean((errors >= 5) & (errors < 10))),
            ">10": float(np.mean(errors >= 10))
        }
    else:
        metrics["error_distribution"] = {"5-10":0, ">10":0}

    return metrics

if __name__=="__main__":
    batches = ["frames-batch-1","frames-batch-2","frames-batch-4","frames-batch-5"]
    stats = init_global_stats()
    global_mistakes = {}

    for b in sorted(batches):
        print("Processing", b)
        run_analysis(b, stats, global_mistakes)

    final_metrics = compute_final_metrics(stats)
    with open("global_metrics_summary.json","w") as f:
        json.dump(final_metrics,f,indent=4)
    with open("global_mistakes_detailed.json","w") as f:
        json.dump(global_mistakes,f,indent=4)
    print("\nDONE")
```
