```python
import os
import cv2
import json
import mediapipe as mp
```

```python
INPUT_DIR = r"frames-real-test-batch\frames"
OUTPUT_DIR = r"frames-real-test-batch\frames_mediapipe"
JSON_PATH = r"frames-real-test-batch\mediapipe_landmarks.json"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
# Mediapipe setup
mp_holistic = mp.solutions.holistic
mp_drawing = mp.solutions.drawing_utils

holistic = mp_holistic.Holistic(
    static_image_mode=True,
    model_complexity=1,
    min_detection_confidence=0.5
)
```

```python
# Prepare JSON dictionary
all_landmarks = {}

# Get image files
image_files = [f for f in os.listdir(INPUT_DIR) if f.lower().endswith((".jpg", ".png", ".jpeg"))]
print(f"Found {len(image_files)} images")
```

```python
for idx, img_file in enumerate(image_files):
    img_path = os.path.join(INPUT_DIR, img_file)
    output_path = os.path.join(OUTPUT_DIR, img_file)

    image = cv2.imread(img_path)
    if image is None:
        print(f"Failed to read {img_file}")
        continue

    # Convert to RGB
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Process with MediaPipe
    results = holistic.process(rgb)

    # Draw hands with colors
    if results.left_hand_landmarks:
        mp_drawing.draw_landmarks(
            image,
            results.left_hand_landmarks,
            mp_holistic.HAND_CONNECTIONS,
            mp_drawing.DrawingSpec(color=(255,0,0)),      # Blue for left hand
            mp_drawing.DrawingSpec(color=(255,0,0))
        )

    if results.right_hand_landmarks:
        mp_drawing.draw_landmarks(
            image,
            results.right_hand_landmarks,
            mp_holistic.HAND_CONNECTIONS,
            mp_drawing.DrawingSpec(color=(255,0,255)),    # Purple for right hand
            mp_drawing.DrawingSpec(color=(255,0,255))
        )

    # Add simple legend
    cv2.putText(image, "Left Hand (Blue)", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,0,0), 2)
    cv2.putText(image, "Right Hand (Purple)", (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,0,255), 2)

    # Save the processed image
    cv2.imwrite(output_path, image)

    # Save landmarks to JSON
    all_landmarks[img_file] = {"left_hand": [], "right_hand": []}

    if results.left_hand_landmarks:
        for lm in results.left_hand_landmarks.landmark:
            all_landmarks[img_file]["left_hand"].append({
                "x": lm.x,
                "y": lm.y,
                "z": lm.z
            })

    if results.right_hand_landmarks:
        for lm in results.right_hand_landmarks.landmark:
            all_landmarks[img_file]["right_hand"].append({
                "x": lm.x,
                "y": lm.y,
                "z": lm.z
            })
```

```python
# Write JSON file
with open(JSON_PATH, "w") as f:
    json.dump(all_landmarks, f, indent=4)

holistic.close()
print(f"Done processing all frames! Landmarks saved to {JSON_PATH}")
```
