```python
import pandas as pd
import re
from pathlib import Path

# Read CSV
csv_path = Path(r"E:\500 newest\500_gloss.csv")
df = pd.read_csv(csv_path)

# Read txt file
with open(r"C:\Users\marha\Downloads\removed-classes.txt", "r") as f:
    text = f.read()

# Extract words between quotes
removed_classes = re.findall(r"'(.*?)'", text)

print("Classes to remove:")
print(removed_classes)

print("Count in txt:", len(removed_classes))

# Remove rows whose gloss exists in txt list
new_df = df[~df["gloss"].isin(removed_classes)]

# Save new csv
output_path = Path(r"E:\500 newest\gloss_filtered.csv")
new_df.to_csv(output_path, index=False)

# Print counts
print("Original CSV count:", len(df))
print("Remaining count:", len(new_df))
print("Removed count:", len(df) - len(new_df))

print("Saved to:", output_path)
```
