```python
import os
import re
import pandas as pd
from collections import defaultdict

FOLDER = r"cleaned_landmarks_147words"

def extract_gloss(filename):
    # remove extension
    name = os.path.splitext(filename)[0]

    # split by spaces
    parts = name.split()

    # CASE 1: last part is a number → instance index → remove it
    if parts and parts[-1].isdigit():
        parts = parts[:-1]

    # rebuild gloss
    gloss = " ".join(parts)

    # normalize separators
    gloss = gloss.replace("_", " ").replace("-", " ")

    # clean extra spaces
    gloss = re.sub(r"\s+", " ", gloss).strip()

    return gloss


counts = defaultdict(int)

for file in os.listdir(FOLDER):
    if file.endswith(".npy"):
        gloss = extract_gloss(file)
        if gloss:  # safety
            counts[gloss] += 1

# convert to dataframe
df = pd.DataFrame([
    {"gloss": k, "count": v}
    for k, v in sorted(counts.items())
])

# save
out_path = os.path.join(FOLDER, "gloss_counts.csv")
df.to_csv(out_path, index=False)

print(df.head())
print("Saved to:", out_path)
```
