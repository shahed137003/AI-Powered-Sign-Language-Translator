```python
#100 words preprocessed and clean no mask
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
from sklearn.metrics import classification_report, accuracy_score

# ============================================================
# 1. CONFIGURATION & PATHS
# ============================================================
DATA_DIR = "/kaggle/input/datasets/mariamhany44/100-words-preprocessed-landmarks-cleaned-no-mask/splits"
CSV_PATH = os.path.join(DATA_DIR, "splits.csv")

MAX_LEN = 96
BATCH_SIZE = 32
LR = 4e-4
EPOCHS = 150
PATIENCE = 30
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ============================================================
# 2. DATASET ENGINE
# ============================================================
class SLRDataset(Dataset):
    def __init__(self, df, aug=False):
        self.df = df.reset_index(drop=True)
        self.aug = aug

    def __len__(self):
        return len(self.df)

    def apply_aug(self, x):
        if self.aug:
            if np.random.rand() < 0.4:
                x += np.random.normal(0, 0.002, x.shape)
            if np.random.rand() < 0.3:
                x *= np.random.uniform(0.95, 1.05)
        
        if x.shape[0] > MAX_LEN:
            x = x[:MAX_LEN]
        else:
            x = np.pad(x, ((0, MAX_LEN - x.shape[0]), (0, 0)), mode='edge')
        return x.astype(np.float32)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        x = np.load(row["filepath"]).astype(np.float32)
        return torch.from_numpy(self.apply_aug(x)), torch.tensor(row["label_id"], dtype=torch.long)

# ============================================================
# 3. V8 ARCHITECTURE (ATTENTION-RESIDUAL)
# ============================================================
class ResidualBlock(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv1d(in_ch, in_ch, 5, padding=2, groups=in_ch),
            nn.Conv1d(in_ch, out_ch, 1),
            nn.BatchNorm1d(out_ch),
            nn.SiLU(),
            nn.Dropout(0.3)
        )
        self.shortcut = nn.Conv1d(in_ch, out_ch, 1) if in_ch != out_ch else nn.Identity()

    def forward(self, x):
        return self.conv(x) + self.shortcut(x)

class SLRModelV8(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.stem = nn.Linear(438, 256)
        self.encoder = nn.Sequential(
            ResidualBlock(256, 256),
            ResidualBlock(256, 512),
            ResidualBlock(512, 512)
        )
        self.attn = nn.Sequential(
            nn.Linear(512, 128),
            nn.SiLU(),
            nn.Linear(128, 1),
            nn.Softmax(dim=1)
        )
        self.classifier = nn.Sequential(
            nn.Linear(512, 512),
            nn.BatchNorm1d(512),
            nn.SiLU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        x = self.stem(x).permute(0, 2, 1) 
        x = self.encoder(x) 
        weights = self.attn(x.permute(0, 2, 1)) 
        features = torch.sum(x * weights.permute(0, 2, 1), dim=-1) 
        return self.classifier(features)

# ============================================================
# 4. PREPARE DATA LOADERS
# ============================================================
df = pd.read_csv(CSV_PATH)
df["filepath"] = df["filepath"].apply(lambda x: os.path.join(DATA_DIR, x.replace("\\", "/")))
labels = sorted(df["label"].unique())
label2id = {l: i for i, l in enumerate(labels)}
df["label_id"] = df["label"].map(label2id)

train_loader = DataLoader(SLRDataset(df[df["split"]=="train"], aug=True), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(SLRDataset(df[df["split"]=="val"]), batch_size=BATCH_SIZE)
test_loader = DataLoader(SLRDataset(df[df["split"]=="test"]), batch_size=BATCH_SIZE)

# ============================================================
# 5. INITIALIZATION & SCHEDULER
# ============================================================
model = SLRModelV8(len(labels)).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=0.05)

# OneCycleLR policy
scheduler = torch.optim.lr_scheduler.OneCycleLR(
    optimizer, max_lr=1e-3, steps_per_epoch=len(train_loader), 
    epochs=EPOCHS, pct_start=0.3
)

criterion = nn.CrossEntropyLoss(label_smoothing=0.15)
scaler = GradScaler()

history = {'tr_l': [], 'tr_a': [], 'vl_l': [], 'vl_a': []}
best_acc = 0
early_stop_cnt = 0

# ============================================================
# 6. TRAINING LOOP
# ============================================================
for epoch in range(EPOCHS):
    model.train()
    tr_loss, tr_corr, tr_total = 0, 0, 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    
    for x, y in pbar:
        x, y = x.to(DEVICE), y.to(DEVICE)
        optimizer.zero_grad()
        with autocast():
            out = model(x)
            loss = criterion(out, y)
        
        scaler.scale(loss).backward()
        
        # Gradient Clipping
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        
        scaler.step(optimizer)
        scaler.update()
        scheduler.step()
        
        tr_loss += loss.item() * x.size(0)
        tr_corr += (out.argmax(1) == y).sum().item()
        tr_total += y.size(0)

    # --- VALIDATION ---
    model.eval()
    vl_loss, vl_corr, vl_total = 0, 0, 0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            out = model(x)
            loss = criterion(out, y)
            vl_loss += loss.item() * x.size(0)
            vl_corr += (out.argmax(1) == y).sum().item()
            vl_total += y.size(0)

    metrics = {
        'tr_l': tr_loss/tr_total, 'tr_a': tr_corr/tr_total,
        'vl_l': vl_loss/vl_total, 'vl_a': vl_corr/vl_total
    }
    for k, v in metrics.items(): history[k].append(v)

    print(f"Epoch {epoch+1}: Tr_Loss: {metrics['tr_l']:.4f} | Tr_Acc: {metrics['tr_a']:.4f} | "
          f"Vl_Loss: {metrics['vl_l']:.4f} | Vl_Acc: {metrics['vl_a']:.4f} | LR: {scheduler.get_last_lr()[0]:.6f}")

    if metrics['vl_a'] > best_acc:
        best_acc = metrics['vl_a']
        torch.save(model.state_dict(), "hail_mary_v8.pt")
        early_stop_cnt = 0
    else:
        early_stop_cnt += 1
        if early_stop_cnt >= PATIENCE: break

# ============================================================
# 7. FINAL EVALUATION
# ============================================================
print("\n--- LOADING BEST MODEL FOR FINAL TEST ---")
model.load_state_dict(torch.load("hail_mary_v8.pt"))
model.eval()
test_preds, test_targets = [], []

with torch.no_grad():
    for x, y in test_loader:
        out = model(x.to(DEVICE))
        test_preds.extend(out.argmax(1).cpu().numpy())
        test_targets.extend(y.numpy())

print(f"\n🎯 FINAL TEST ACCURACY: {accuracy_score(test_targets, test_preds):.4f}")
print("\nREPORT:\n", classification_report(test_targets, test_preds, target_names=labels))

# Plot Results
fig, ax = plt.subplots(1, 2, figsize=(15, 5))
ax[0].plot(history['tr_l'], label='Train Loss')
ax[0].plot(history['vl_l'], label='Val Loss')
ax[0].set_title("Loss History")
ax[1].plot(history['tr_a'], label='Train Acc')
ax[1].plot(history['vl_a'], label='Val Acc')
ax[1].set_title("Accuracy History")
plt.legend()
plt.show()
```
