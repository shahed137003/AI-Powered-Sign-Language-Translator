```python
import pandas as pd

# Load your final dataset
df = pd.read_csv("final_gloss_files.csv")

# =========================
# Filter classes < 25 samples
# =========================
less_than_25 = df[df["count"] < 25]

# =========================
# Results
# =========================
print("===== Low Sample Classes (< 25 instances) =====")
print("Number of classes:", len(less_than_25))
print("Total classes:", len(df))

print("\nPercentage:", round(len(less_than_25) / len(df) * 100, 2), "%")

# =========================
# Show actual classes
# =========================
print("\n===== Glosses with < 25 samples =====")
print(less_than_25[["gloss", "count"]].sort_values("count"))
```
